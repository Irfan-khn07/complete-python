```python
# write a program , ask the user to enter phone number do sum till single digit
# Program to find the single digit (digital root) of a phone number

phone = input("Enter your phone number: ")

# Keep only digits in case the user enters +91, spaces, or hyphens
digits = ''.join(ch for ch in phone if ch.isdigit())

while len(digits) > 1:
    total = 0
    for digit in digits:
        total += int(digit)
    print("Sum =", total)
    digits = str(total)

print("Single Digit =", digits)
```

## Break
- it will stop the loop


```python
l = ['raj','khhana','yesh','kumar','sivam']
for i in l:
    if i =='yesh':
        break
    print(i)
```

## Continue
it will continue with next items

```python
l = ['raj','khhana','yesh','kumar','sivam']
for i in l:
    if i =='yesh':
        continue
    print(i)
```
