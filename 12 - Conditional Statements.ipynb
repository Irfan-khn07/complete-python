```python
a,b,c,d=2,2,3,4  
if (a==b and c!=d): # both condition is true
    print('abc')
else:
    print('def')

print("khan")
```

## elif
**Syntax**
if condition:

    if block of code

elif condition:
    
    block of code
- if block condition is True  , if block will be executed
- if block condition is False  , it will check elif block condition
- elif block condition is True  , elif block will be executed
- elif block condition is False  , elif block will be ignored

```python
# Q. ask user to inter marks if marks are greater 90, excellent, if marks are greater or equal to 70 , good if marks are less then 70, faild

marks = eval(input('inter your marks'))
if marks>90:
    print('excellent')
elif marks>=70:
    print('good')
elif marks<70:
    print('faild')
```

### Nested if 
**- within if one more if is called nested if**

- We can have a if .. elif...else statement inside another if ...elif...else statement. this is called nesting in computer programming.
### Syntax
if condition:

    if condition:
        block of code

```python
if 2>3:
    print('hi')
    if 2>3:
        print('bye')
else:
    print('im kareli')
```

```python
if 2<3:
    print('hi')
    if 2>3:
        print('bye')
else:
    print('im kareli')
```
