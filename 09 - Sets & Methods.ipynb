## Set Methods
**set.add()**

**- add** methods adds an one item to the set. if the item already exists in the set, nothing will happens

```python
s = {1,2,3}
print(s)
print(id(s))

s.add(4) # add value
print(s)
print(id(s))
```

**set.update()**

**- update** methods updates the set with another set or any iterable

```python
s1 = {1,2,3}
s1.update((4,5,6))
print(s1)
print(id(s1))
```

```python
s2 = {1,2,3}
s2.add((4,5,6))
print(s2)
```

**set.discard()**

- **discard** method removes the specified item from the set. if the item dose not exist in the set, nothing happens

```python
s = {1,2,3}
s.discard(3)
print(s)
```

```python
s = {1,2,3}  #out range
s.discard(5 )
print(s)
```

**t.remove()**

- **removes** method removes the specified element from the set. if the element dose not exist in the set, a keey error is raised.


```python
s = {1,2,3,4}
s.remove(3)
print(s)
```

**Srt.clear()**

- **clear** methods remove all elements from the set

```python
s = {1,2,3,4}
s.clear() # removes all items in the list
print(s)
```

```python
s = {1,2,3}
del s            #del the variable
print(s)
```

**set.copy()**

- **copy** method returns a shallow copy of the list
- changes to copied set donen't affect the original set

```python
s1 = {1,2,3}
s2 = s1.copy()
s2.add(4)
print(s1)
print(s2)
```

```python
s = {12,5,8,"srk",(1,2,3),34,6}
s.pop()  # remove one value
print(s)
```

**set.intersection()** ------ return comman element
  
- **difference** methods returns a new set with elements that are common to all set

```python
A = {1,2,3}
B = {2,3,4}

A.intersection(B)
```

**set.difference()**

- A U B = B U A , A intersection B = B itersection A ,but A - B =/ B -A
- **difference** method returns a new set with elements in the set that are not in the other specified sets

```python
setA = {1,2,3}
setB = {2,3,4}

print(setA - setB) # within setA remove the setB items
print(setB - setA) # within setB remove the setA items
```

 **set.symmetric_difference()**

 - **symmetric_difference()** method return a new set with elements that are either of the sets, but not in both
 - (A union B) -(A intersection b)

```python
setA = {1,2,3,"khan"}
setB = {2,3,4}

setA.symmetric_difference(setB) # with union remove intersection items
```
