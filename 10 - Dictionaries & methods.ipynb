- Each item is key:value
- Each value is stored in the respective keys
- keys can be immutable(int, float, bool, str, tuple)
- value can be anything (int, float, bool, str, tuple, list, set, dict)
- if any 1 value is dict, then it is called Nested dictionary
- Dictionary is Mutable (where you can add, remove, replace)
- Indexing
- Dictionary Methods
  - items()
  - keys()
  - value()
  - update()
  - pop()
  - clear()
  - replace the value
  - repalce the item

### Dictionary Methods
**dict.items()**

- **items()** methods return a list of key value pairs in the dictionary
- list of tuple
- each tuple consists of (key, value)

```python
marks.items() #tuple
```

**dict.keys()**

- **keys()** methods returns a list of keys in the dict

```python
marks.keys() #list
```

**dict.values()**

- **values()** method returns a list of value in the dict


```python
marks.values()
```

**To add asingle item into a dic**

- option 1

```python
print(marks)
marks.update({'Hindi':90})
print(marks)
```

```python
#secd
marks["Hindi"] = 70 , 43
print(marks)
```

**To add multipal items into a dict**


```python
marks1 = {'history':24 , 'hindi': 53}
marks1.update({'chemistry':89 , 'Maths': 58})
print(marks1)
```

**To delete items in dict**

- option 1: del dict[key]

```python
del marks1['Maths']
print(marks1)
```

- option 2 : **dict.pop(key)**
- **pop()** method removes and return the value of the specified key.
- if the key dose not exist , it raises a keyError
- one item remove

```python
marks = {'history': 43, 'Hindi':54}
marks.pop("history")
marks
```

**dict.clear()**

- **clear()** methods remove all items from dic

```python
marks = {'history': 43, 'Hindi':54}
marks.clear()
marks
```

**dict.copy()**

- copy() methods returns a shallow copy of the dic
- changes to the copy of the dict not affect the original dic

```python
d = {'history': 49, 'hindi': 58}
a = d.copy()    # d ke value ko a me copy kiya
a.update({'math':89})
print(a)
print(d)
```

**dict.get()**

- **get()** Methods returns the value of the specified key.
- if the key is not present in the dict, it will return "None"
- we can get the value by using another way also **dict_name[key]**

```python
A = {'history': 49, 'hindi': 58}
A.get("hindi")
```
