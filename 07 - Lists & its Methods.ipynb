## list.extend()
- used to add multiple items in a list
- it adds each item individually at the end

**Whateven items you want to add, take them into a list & apply extend()**

```python
l = [12,33,5,7,6,75]
l.append([46,57,324])
print(l)
```

```python
l = [12,33,5,7,6,75]
l.extend([46,57,324]) # multipla 
print(l)
```

```python
l1 = [10,20,30]
l2 = [12,23,563]
# l1 = l1 + l2 cancatination both are same
l1.extend(l2)
print(l1)
```

**list.sort()**

- first preoraty is Capslock
- used to sort items in a list either in asending order or descending order
- by default its sort in ascending order
- **Note** Sort is applicable for either only alphabet or only numeric values

```python
l = [1,23,32,3.1]
l.sort()   # Ascending order
l
```

```python
l = ["siva","srk","rama"]
l.sort() #str
print(l)
```

**list.copy()**

- used to create a new object with copy of items
- **Deep copy** -- all datatype
- **Shallow copy** -- for list

```python
a = [1,2,3]
b = a                  # deep copy
b.append(4)
print(a)
print(b)
id(b) , id(a)
```

```python
a = [1,2,3]
b = a.copy()                # Shallow copy
b.append(4)
print(a)
print(b)
id(b) , id(a)
```
