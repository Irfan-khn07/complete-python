```python
# Q. Select any 10 num between 1 to 100 & store them in a list
import random
l = []
for i in range(10):
    l.append(random.randint(1,100))
print(l)
             
```

```python
# Q. Select any 10 num between 1 to 100 without any repeation & store them in a list
import random
l = []
for i in range(1000000):
    if i not in l and len(l)<=10:
        l.append(random.randint(1,100))

print(l)
```

```python
# Q. Select any 10 even num between 1 to 100 without any repeation & store them in a list
import random
l = []
for i in range(1000000):
    a = random.randint(1,100)
    if a not in l and len(l)<10 and a%2==0:  # if i want only those no who divided by 5 then # a%5==0
        l.append(a)

print(l)
```

```python
# ex with while loop of 5
l = []
while True:
    a = random.randint(1,100)
    if a not in l and a%5==0:
        l.append(a)
    if len(l)==10:
        break
print(l)
```

## Modify the values in Numpy
- Add or insert
- Remove or delete
- Replace

**Add a single values in a array**

- Syntax : np.append(array,values)

```python
a = np.array([1,4,5,7,8,9])

a1 = np.append(a,786)
print(a1)
```

**Add a maltipal values in a array**
- Syntax : np.append(array,list of values)

```python
a = np.array([1,4,5,7,8,9])

a2 = np.append(a,[2,6,12,45,76])
print(a2)
```

**Delete a single values in a array**

- Syntax : np.delete(array,index numbers)

```python
a = np.array([1,4,5,7,8,9])

a1 = np.delete(a,3)
print(a1)
```

**Delete a multipal values in a array**

- Syntax : np.delete(array,[multipal index numbers])

```python
a = np.array([1,4,5,7,8,9])

a1 = np.delete(a,[2,5,1])
print(a1)
```

**Replacing a single value in a 1D array**

```python
a = np.array([1,4,5,7,8,9])

a[1] = 12
print(a)
```

**Replacing a multiple value with a single values in a array**

```python
a = np.array([1,4,5,7,8,9])

a[[0,2,4]] = 20
print(a)
```

**Replacing a multiple value with a multiple values in a array**

```python
a = np.array([1,4,5,7,8,9])

a[[1,2,3,5]] = [11,12,13,14]
print(a)
```

**Copy array**

```python
a = np.array([1,2,3,4])
b = a.copy()
b[0] = 100
print(a)
print(b)    # b me 1 ---- 100 ho jayega
```

**Sorting an Array**

```python
a = np.array([100,4,6,2,1,10])
print(a)

b = np.sort(a)
print(b)
```
